```python
#carrega biblioteca
from transformers import pipeline

#função generica para não ter que escrever sempre a mesma coisa
def rodar_pipeline(tarefa, modelo, entrada, **kwargs):
    executor = pipeline(tarefa, model=modelo)
    return executor(entrada, **kwargs)
```

```python
#teste básico do gerador de texto

frase_inicial = "O futuro da inteligência artificial no Brasil depende de"

resultado = rodar_pipeline(
    "text-generation",
    "pierreguillou/gpt2-small-portuguese",
    frase_inicial,
    max_length=60,
    do_sample=True,
)

print(resultado[0]["generated_text"])
```

```python
#Controlando a criatividade do modelo

gerador = pipeline("text-generation", model="pierreguillou/gpt2-small-portuguese")

texto_greedy = gerador(frase_inicial, max_length=40, do_sample=False)
texto_criativo = gerador(frase_inicial, max_length=40, do_sample=True, temperature=1.2, top_k=50)

print("Greedy (determinístico):\n", texto_greedy[0]["generated_text"])
print("\nSampling (mais criativo):\n", texto_criativo[0]["generated_text"])
```

```python
#preenchimento de lacunas

frases = [
    "O jogador chutou a bola em direção ao [MASK].",
    "São Paulo é a maior [MASK] do Brasil.",
]

sugestoes = rodar_pipeline(
    "fill-mask",
    "neuralmind/bert-base-portuguese-cased",
    frases,
)

for frase, opcoes in zip(frases, sugestoes):
    print(f"Frase: {frase}")
    melhor = opcoes[0]
    print(f"  Melhor sugestão: {melhor['token_str']} (score={melhor['score']:.3f})")
    print()
```

```python
#analise de sentimentos

avaliacoes = [
    "Adorei o produto, chegou rápido e funciona perfeitamente!",
    "Péssimo atendimento, não recomendo a ninguém.",
    "O produto é ok, cumpre o que promete.",
]

resultados = rodar_pipeline(
    "sentiment-analysis",
    "cardiffnlp/twitter-xlm-roberta-base-sentiment",
    avaliacoes,
)

for texto, r in zip(avaliacoes, resultados):
    print(f"{texto}\n  -> {r['label']} (score={r['score']:.3f})\n")
```

```python
#classificação zero shot
texto = "O time perdeu de virada nos últimos minutos da partida."
categorias = ["futebol", "política", "tecnologia"]

resultado = rodar_pipeline(
    "zero-shot-classification",
    "joeddav/xlm-roberta-large-xnli",
    texto,
    candidate_labels=categorias,
)

for rotulo, score in zip(resultado["labels"], resultado["scores"]):
    print(f"{rotulo}: {score:.3f}")
```
