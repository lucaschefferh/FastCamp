```python
#Carrega biblioteca e modelo
import torch
from transformers import AutoTokenizer, AutoModelForQuestionAnswering

model_name = "pierreguillou/bert-base-cased-squad-v1.1-portuguese"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForQuestionAnswering.from_pretrained(model_name)
```

```python
#Passa texto e tokeniza ambos
texto = "Carl Sagan foi um cientista norte-americano. Sagan é autor de mais de 600 publicações científicas e também de mais de vinte livros de ciência e ficção científica."
pergunta = "Carl Sagan é um cientista norte-americano"

#tokenização do texto e da pergunta
inputs = tokenizer(pergunta, texto, return_tensors="pt")
```

```python
#inferência direta pelo PyTorch
with torch.no_grad():
    outputs = model(**inputs)

#ncontra a posição do início e fim da resposta nos logits
start_idx = torch.argmax(outputs.start_logits)
end_idx = torch.argmax(outputs.end_logits) + 1

#ecodifica os tokens para texto
resposta = tokenizer.decode(inputs["input_ids"][0][start_idx:end_idx])
print("Resposta:", resposta)
```
