```python
#fazendo as importações
import transformers
from transformers import pipeline
```

```python
#Carregando o modelo dentro do pipeline
mascarar = pipeline("fill-mask", model="neuralmind/bert-base-portuguese-cased")
```

```python
#Passando o texto para dentro do pipeline e printando as mais provaveis palavras
texto = mascarar("Existe uma chance do copo cair no [MASK]")
for x in range(len(texto)):
  print(texto[x])
```

```python
#Memsa coisa para duas sentenças diferentes
texto = mascarar(["Existe uma chance do copo cair no [MASK]","Brasília é a capital do [MASK]"])
for x in range(len(texto)):
  print(texto[x])
```
