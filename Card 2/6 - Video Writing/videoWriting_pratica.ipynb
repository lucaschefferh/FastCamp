```python
import os
import cv2
import matplotlib.pyplot as plt
from IPython.display import YouTubeVideo, display, HTML
from base64 import b64encode
%matplotlib inline
```

```python
#lendo e recebendo os frames do vídeo
video = cv2.VideoCapture("video_aleatorio.mp4")
ret, frame = video.read()
```

```python
#plotando o primeiro frame em rgb
plt.imshow(frame[..., ::-1])
```

```python
#recebendo a altura e a largura do vídeo
frame_largura = int(video.get(3))
frame_altura = int(video.get(4))

#passando o video que temos para formato .avi
saida_avi = cv2.VideoWriter("video_aleatorio.avi", cv2.VideoWriter_fourcc("M", "J", "P", "G"), 10, (frame_largura, frame_altura))
```

```python
#Agora passando para o objeto criado frame a frame do vídeo 
while video.isOpened():
    ret, frame = video.read()

    if ret:
        saida_avi.write(frame)
    else:
        break
```

```python
#Por fim liberando ambos os vídeos
video.release()
saida_avi.release()
```
