```python
import cv2
import glob
import os
import sys
```

```python
#nome da pasta e extensão dos arquivos
pasta_imagens = "Crossing/img" 
extensao = "*.jpg" 

#busca e ordena todos os frames
caminhos = sorted(glob.glob(os.path.join(pasta_imagens, extensao)))

#le o primeiro frame
frame_atual = cv2.imread(caminhos[0])

#cria o rastreador CRST
rastreador = cv2.legacy.TrackerCSRT_create()
```

```python
#abre a janela no primeiro frame
caixa_selecao = cv2.selectROI("Selecione o objeto e aperte enter", frame_atual, False)

#inicializa o rastreador baseado na seleção
rastreador.init(frame_atual, caixa_selecao)

#destroi a janela quando apertado enter
cv2.destroyWindow("Selecione o objeto e aperte enter")
```

```python
#a partir do primeiro frame começa ler
for i in range(1, len(caminhos)):
    frame_atual = cv2.imread(caminhos[i])
    
    if frame_atual is None:
        break

    #atualiza o rastreio
    sucesso, caixa = rastreador.update(frame_atual)

    #se tiver sucesso desenha um retangulo em verde com os frames
    if sucesso:
        (x, y, w, h) = [int(v) for v in caixa]
        cv2.rectangle(frame_atual, (x, y), (x + w, y + h), (0, 255, 0), 2)
        cv2.putText(frame_atual, f"Frame: {i}", (20, 30), 
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2)
    #caso contrario escreve objeto perdido
    else:
        cv2.putText(frame_atual, "objeto perdido", (20, 30), 
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 0, 255), 2)

    cv2.imshow("Rastreamento", frame_atual)

    #telca q ou esc para sair
    if cv2.waitKey(30) & 0xFF == ord('q'):
        break

cv2.destroyAllWindows()
print("Rastreamento concluído")
```
