```python
#carregamento do modelo
protoFile   = "pose_deploy_linevec_faster_4_stages.prototxt"
weightsFile = os.path.join("model", "pose_iter_160000.caffemodel")
```

```python
#variaveis para a deteccao de pose
nPoints = 15
POSE_PAIRS = [
    [0, 1],
    [1, 2],
    [2, 3],
    [3, 4],
    [1, 5],
    [5, 6],
    [6, 7],
    [1, 14],
    [14, 8],
    [8, 9],
    [9, 10],
    [14, 11],
    [11, 12],
    [12, 13],
]

#carregando o modelo
net = cv2.dnn.readNetFromCaffe(protoFile, weightsFile)
```

```python
#pegando as informações da imagem a ser utilizada
im = cv2.imread("Tiger_Woods_crop.png")
im = cv2.cvtColor(im, cv2.COLOR_BGR2RGB)

inWidth  = im.shape[1]
inHeight = im.shape[0]
```

```python
Image(filename="Tiger_Woods.png")
```

```python
#passando o tamanho da imagem e criando o blob
netInputSize = (368, 368)
inpBlob = cv2.dnn.blobFromImage(im, 1.0 / 255, netInputSize, (0, 0, 0), swapRB=True, crop=False)
net.setInput(inpBlob)
```

```python
#recebdno a saida da rede neural
output = net.forward()

#mostrando o mapa de probabilidade
plt.figure(figsize=(20, 5))
for i in range(nPoints):
    probMap = output[0, i, :, :]
    displayMap = cv2.resize(probMap, (inWidth, inHeight), cv2.INTER_LINEAR)
    
    plt.subplot(2, 8, i + 1)
    plt.axis("off")
    plt.imshow(displayMap, cmap="jet")
```

```python
#escala de x e y
scaleX = inWidth  / output.shape[3]
scaleY = inHeight / output.shape[2]

#lista vazia para guardas os pontos detectados
points = []

#limiar 
threshold = 0.1

for i in range(nPoints):
    #recebe o mapa de probabilidade
    probMap = output[0, i, :, :]

    #acha o minimo local do mapa
    minVal, prob, minLoc, point = cv2.minMaxLoc(probMap)

    #escala para por no mapa
    x = scaleX * point[0]
    y = scaleY * point[1]

    if prob > threshold:
        #adiciona a lista se a probabilidade for maior que o limiar
        points.append((int(x), int(y)))
    else:
        points.append(None)
```

```python
imPoints = im.copy()
imSkeleton = im.copy()

# desenha os pontos enumerados
for i, p in enumerate(points):
    cv2.circle(imPoints, p, 8, (255, 255, 0), thickness=-1, lineType=cv2.FILLED)
    cv2.putText(imPoints, "{}".format(i), p, cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 0, 0), 2, lineType=cv2.LINE_AA)

#desenha e liga os pontos
for pair in POSE_PAIRS:
    partA = pair[0]
    partB = pair[1]

    if points[partA] and points[partB]:
        cv2.line(imSkeleton, points[partA], points[partB], (255, 255, 0), 2)
        cv2.circle(imSkeleton, points[partA], 8, (255, 0, 0), thickness=-1, lineType=cv2.FILLED)
```

```python
#plota ambas imagens
plt.figure(figsize=(50, 50))

plt.subplot(121)
plt.axis("off")
plt.imshow(imPoints)

plt.subplot(122)
plt.axis("off")
plt.imshow(imSkeleton)
```
