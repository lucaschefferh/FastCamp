```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Image
```

```python
#lendo a imagem em escala de cinza e plotando
escudo = cv2.imread("maiorDoMundo.jpg", 0)

plt.imshow(escudo, cmap='gray')
```

```python
escudo.shape
```

```python
#trocando os primeiros pixels para preto
editado = escudo.copy()
editado[0:300, 0:300] = 0
plt.imshow(editado, cmap='gray')
```

```python
bgr = cv2.imread("maiorDoMundo.jpg", cv2.IMREAD_COLOR)
rgb = bgr[:, :, ::-1]

plt.imshow(rgb)
```

```python
#tentando dar um zoom na imagem
zoom = rgb[50:670, 300:900]
plt.imshow(zoom)
```

```python
zoom2 = rgb[25:690, 300: 970]
plt.imshow(zoom2)
cv2.imwrite("escudo_ajustado.jpg", zoom2)
```

```python
imagem_cinza = cv2.imread("escudo_ajustado.jpg", 0)
imagem = cv2.imread("escudo_ajustado.jpg")
imagem_invertida = cv2.flip(imagem, 0)
```

```python
plt.figure(figsize=(18, 5))
plt.subplot(141);plt.imshow(imagem_cinza, cmap = 'gray');plt.title("Imagem em escala de cinza");
plt.subplot(142);plt.imshow(imagem);plt.title("Imagem em RGB");
plt.subplot(143);plt.imshow(imagem_invertida);plt.title("Imagem invertida");
```
