```python
imagem = cv2.imread("images.jpeg")
print(imagem)
print("Altura e largura: ", imagem.shape)
```

```python
#le as imagens em escala de cinza e inverte de bgr para rgb
img_gray = cv2.imread("images.jpeg", 0)
img_rgb = imagem[:, :, ::-1]
```

```python
plt.figure(figsize=[20,5])
plt.subplot(141);plt.imshow(imagem);plt.title("Imagem em BGR");
plt.subplot(142);plt.imshow(img_gray, cmap='gray');plt.title("Imagem em escala de cinza");
plt.subplot(143);plt.imshow(img_rgb, cmap='gray');plt.title("Imagem em RGB");
```
