```python
import cv2
import matplotlib
import numpy as np
import matplotlib.pyplot as plt
```

```python
#lendo a imagem e passando para rgb 
imagem = cv2.imread("maiorDoMundo.jpg")
imagem_rgb = imagem[:, :, ::-1]
plt.imshow(imagem_rgb)
```

#### Feita a cópia da imagem, passado os parametros do texto e plotado a imagem junto do texto

```python
imagemFinal = imagem_rgb.copy()

text = "O maior time brasileiro -->"
fontScale = 2.0 
fontFace = cv2.FONT_HERSHEY_PLAIN 
fontColor = (0, 0, 0) 
fontThickness = 2 
 
cv2.putText(imagemFinal, text, (10, 70), fontFace, fontScale, fontColor, fontThickness, cv2.LINE_AA);
#plotando a imagem
plt.imshow(imagemFinal)
```
